```python
## Importing library and Reading CSV file

import pandas as pd
import matplotlib.pyplot as plt

netflix_df = pd.read_csv('netflix_data.csv')
netflix_df.head()
```

```python
## Subsetting the Release year Data

rel_year = netflix_df['release_year'].between(1990,1999)
netflix_df = netflix_df[rel_year]
netflix_df['release_year'].value_counts()                               # value_counts() = it sums the numeric value present in the Row or column
```

```python
## Showing Histogram

plt.hist(netflix_df['release_year'])
plt.title("Distribution of Movies released in 90s")
plt.xlabel('Year')
plt.ylabel('No of Movies')
plt.show()
```

```python
## Task:1 - Identify Most frequent movie duration in 90s

duration = netflix_df['duration'].mode().sum()
print(duration)
print(type(duration))
```

```python
## Subsetting genre from the movie data and Plotting

netflix_df['genre'].value_counts().plot(kind='bar',figsize=(20,6), rot=0)             # figsize = width and height, kind = type of chart , rot = rotate
plt.title("Genre of movies Release in 90s")
plt.xlabel('Genre')
plt.ylabel('No of Movies')
plt.show()
```

```python
## Task:2 - Count Short Action movies in 90s

short_action_movies = (netflix_df["genre"] == "Action") & (netflix_df["duration"] < 90)

movie_count = short_action_movies.sum()
print(movie_count)
print(type(movie_count))
```
